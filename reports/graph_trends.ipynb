```python
import matplotlib.pyplot as plt
import get_open_vsx_admin_reports
from datetime import datetime
import pandas as pd

STARTING_YEAR = 2021
STARTING_MONTH = 11

df = get_open_vsx_admin_reports.get_publishing_data(STARTING_YEAR, STARTING_MONTH)
```

```python
df
```

```python
dates = []
index = 0
df['day'] = 1
df['dates'] = pd.to_datetime(df[['year', 'month', 'day']])

# 3. (Optional) Drop the 'day' column if you no longer need it
df = df.drop(columns=['day'])
df
```

```python
downloads_graph = []
for total in df['downloads']:
    downloads_graph.append(total / 1000000)
df['downloads_graph'] = downloads_graph
```

```python
y_max = df['downloads_graph'][len(df['downloads_graph'])-1]
y_axis_max = int(y_max) + 5
plt.ylim([0, y_axis_max])
# plt.style.use('seaborn-dark')
plt.rcParams["figure.figsize"] = (15,10)
plt.grid(True)
plt.plot(df['dates'], df['downloads_graph'], marker='o')
plt.title('Downloads per Month', fontsize=18)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Downloads (Millions)', fontsize=14)
plt.xticks(fontsize=10)
plt.xticks(rotation=45)
plt.yticks(fontsize=12)
plt.savefig('downloads.png')
plt.show()
df
```

```python
start_limit = pd.to_datetime('2025-05-01')
end_limit = pd.to_datetime('2026-04-30')
plt.xlim(start_limit, end_limit)
y_max = df['downloads_graph'][len(df['downloads_graph'])-1]
y_axis_max = int(y_max) + 5
plt.ylim([0, y_axis_max])
# plt.style.use('seaborn-dark')
plt.rcParams["figure.figsize"] = (15,10)
plt.grid(True)
plt.plot(df['dates'], df['downloads_graph'], marker='o')
plt.title('Downloads per Month', fontsize=18)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Downloads (Millions)', fontsize=14)
plt.xticks(fontsize=10)
plt.xticks(rotation=45)
plt.yticks(fontsize=12)
plt.savefig('downloads.png')
plt.show()
```

```python
y_max = df['extensions'][len(df['extensions'])-1]
y_axis_max = (int(y_max / 1000) + 1) * 1000
plt.ylim([0, y_axis_max])
# plt.style.use('seaborn-dark')
plt.rcParams["figure.figsize"] = (15,10)
plt.grid(True)
plt.plot(df['dates'], df['extensions'], marker='o', label='Extensions')
plt.plot(df['dates'], df['publishers'], marker='o', label='Publishers')
plt.title('Total Published Extensions and Publishers by Month', fontsize=18)
plt.xlabel('Month', fontsize=14)
plt.xticks(fontsize=10)
plt.xticks(rotation=45)
plt.ylabel('Total Extensions/Publishers', fontsize=14)
plt.yticks(fontsize=12)
plt.legend(fontsize=14)
plt.savefig('publications.png')
plt.show()
```
